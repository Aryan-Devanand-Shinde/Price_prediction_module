```python
def reverse_geocode_full(lat, lon):
    import requests
    from requests.exceptions import RequestException

    url = "https://nominatim.openstreetmap.org/reverse"
    params = {
        "format": "jsonv2",
        "lat": lat,
        "lon": lon,
        "zoom": 10,
        "addressdetails": 1
    }
    headers = {"User-Agent": "CropRecommender/1.0"}

    try:
        r = requests.get(url, params=params, headers=headers, timeout=15)
        r.raise_for_status()
        data = r.json()
    except RequestException as e:
        raise RuntimeError(f"Reverse geocode failed: {e}")

    address = data.get("address", {})

    # =========================
    # ✅ STATE
    # =========================
    state = (
        address.get("state") or
        address.get("region") or
        address.get("province") or
        "Unknown"
    )

    # =========================
    # ✅ DISTRICT
    # =========================
    district = (
        address.get("state_district") or
        address.get("county") or
        address.get("district") or
        "Unknown"
    )

    district = district.lower().replace("district", "").strip().title()

    # =========================
    # ✅ CITY (FIXED LOGIC 🔥)
    # =========================
    city_raw = (
        address.get("city") or
        address.get("town") or
        address.get("municipality") or
        address.get("suburb")
    )

    if city_raw:
        city = city_raw.strip().title()
    else:
        # 🔥 fallback to district (IMPORTANT FIX)
        city = district

    return state, district, city
```

```python
STATE, DISTRICT, CITY = reverse_geocode_full(lat, lon)
print(STATE, DISTRICT, CITY)
```
