```python
# =========================
# INSTALL (run once)
# =========================
# !pip install ultralytics

from ultralytics import YOLO

# =========================
# CONFIG
# =========================
DATA_YAML = r"C:\Users\shind\Desktop\BE_PROJECT\PRICE PREDICTION MODEL\mc\apple.yaml"

# =========================
# LOAD MODEL
# =========================
# Use yolov8s for better accuracy (recommended)
model = YOLO("yolov8s.pt")

# =========================
# TRAIN
# =========================
model.train(
    data=DATA_YAML,   # path to apple.yaml
    epochs=50,        # increase to 100 for better results
    imgsz=640,
    batch=16,
    device=0,         # GPU (use 'cpu' if no GPU)
    workers=4,
    name="apple_detector",  # folder name
    pretrained=True,
    optimizer="Adam",
    lr0=0.001
)

# =========================
# DONE
# =========================
print("🔥 Training complete!")
```
